```python

# Import required libraries
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler

# Load Titanic dataset
df = pd.read_csv(r"C:\Users\Tanuja\Downloads\titanic.csv")

# Check dataset columns
print("Dataset Columns:")
print(df.columns.tolist())

# Display original dataset
print("\nOriginal Dataset:")
print(df.head())

# Display dataset information
print("\nDataset Information:")
df.info()

# Check missing values before processing
print("\nMissing Values Before Processing:")
print(df.isnull().sum())

# Handle missing numerical values using mean
df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Fare"] = df["Fare"].fillna(df["Fare"].mean())

# Handle missing categorical values using mode
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])

# Remove unnecessary columns
df.drop(["PassengerId", "Name", "Ticket", "Cabin"],
        axis=1, inplace=True)

# Apply Label Encoding to Sex
le = LabelEncoder()
df["Sex"] = le.fit_transform(df["Sex"])

# Display Label Encoding mapping
print("\nSex Encoding:")
print(dict(zip(le.classes_, le.transform(le.classes_))))

# Apply One-Hot Encoding to Embarked
df = pd.get_dummies(df, columns=["Embarked"], dtype=int)

# Display dataset after encoding
print("\nDataset After Encoding:")
print(df.head())

# Select numerical attributes for scaling
cols = ["Pclass", "Age", "SibSp", "Parch", "Fare"]

# Apply Standard Scaling
scaler = StandardScaler()
df[cols] = scaler.fit_transform(df[cols])

# Display dataset after scaling
print("\nDataset After Scaling:")
print(df.head())

# Check missing values after processing
print("\nMissing Values After Processing:")
print(df.isnull().sum())

# Separate input features and target variable
X = df.drop("Survived", axis=1)
y = df["Survived"]

print("\nInput Features:")
print(X.head())

print("\nTarget Variable:")
print(y.head())

# Display final dataset shape
print("\nFinal Dataset Shape:")
print(df.shape)

# Save cleaned dataset
df.to_csv(
    r"C:\Users\Tanuja\Downloads\cleaned_titanic_dataset.csv",
    index=False
)

print("\nCleaned Dataset Saved Successfully!")
```

```python
# PRACTICAL 1 – FEATURE ENGINEERING

## 1. AIM

To apply basic feature engineering techniques such as handling missing values, encoding categorical variables, and scaling numerical attributes on a suitable open-access dataset to prepare a clean and model-ready dataset for machine learning analysis.

## 2. DATASET USED

**Dataset Name:** Titanic Passenger Dataset

**Dataset Location:** `C:\Users\Tanuja\Downloads\titanic.csv`

### Why is the Titanic dataset used?

The Titanic dataset is selected because it contains different types of data required to demonstrate feature engineering techniques.

1. It contains numerical attributes such as Age, Fare, Pclass, SibSp and Parch.
2. It contains categorical attributes such as Sex and Embarked.
3. It contains missing values in columns such as Age, Cabin and Embarked.
4. It contains a target variable named Survived, which indicates whether a passenger survived or not.
5. It is useful for demonstrating missing value handling, categorical encoding and numerical feature scaling.

Therefore, the Titanic dataset is suitable for this practical because it contains numerical data, categorical data and missing values.

## 3. THEORY

**Feature Engineering** is the process of selecting, creating and transforming raw data into meaningful features that can be effectively used by machine learning algorithms.

Raw datasets may contain missing values, categorical attributes and numerical features with different ranges. These issues must be handled before applying many machine learning algorithms.

The main techniques used in this practical are:

### A. Handling Missing Values

Missing values are data entries that are not available or recorded in a dataset. They are commonly represented by NaN in Python.

Missing values can affect data analysis and the performance of machine learning algorithms.

Different methods used to handle missing values are:

- **Mean Imputation:** Replaces missing numerical values with the average of available values.
- **Median Imputation:** Replaces missing numerical values with the middle value.
- **Mode Imputation:** Replaces missing values with the most frequently occurring value.
- **Deletion:** Removes rows or columns containing missing values.

In this practical, mean imputation is used for the Age and Fare columns, while mode imputation is used for the Embarked column.

**Why are different methods used?**

Numerical attributes contain measurable values, so mean or median can be used. Categorical attributes contain labels or categories, so mode is generally used.

### B. Encoding Categorical Variables

Categorical variables contain values represented by labels or groups instead of numerical measurements.

Most traditional machine learning algorithms require numerical input. Therefore, categorical variables are converted into numerical representations through encoding.

There are two encoding techniques used in this practical.

**1. Label Encoding**

Label Encoding converts categorical values into numerical labels by assigning a unique integer to each category.

Example:

| Sex | Encoded Value |
|---|---|
| Female | 0 |
| Male | 1 |

In this practical, Label Encoding is applied to the Sex column.

**2. One-Hot Encoding**

One-Hot Encoding converts each category into a separate binary column containing 0 or 1.

Example:

| Embarked | Embarked_C | Embarked_Q | Embarked_S |
|---|---:|---:|---:|
| C | 1 | 0 | 0 |
| Q | 0 | 1 | 0 |
| S | 0 | 0 | 1 |

In this practical, One-Hot Encoding is applied to the Embarked column.

### C. Feature Scaling

Feature Scaling is the process of transforming numerical attributes into a comparable scale.

Different attributes may have different numerical ranges. For example, Age and Fare may have very different magnitudes.

Feature scaling prevents features with larger numerical magnitudes from unnecessarily dominating certain machine learning algorithms.

Two commonly used feature scaling techniques are:

**1. Standardization**

Standardization transforms numerical values to approximately zero mean and unit standard deviation.

It is performed using StandardScaler.

Formula:

Z = (X - μ) / σ

Where:

- X = Original value
- μ = Mean of the feature
- σ = Standard deviation
- Z = Standardized value

**2. Normalization**

Normalization converts numerical values into a fixed range, commonly between 0 and 1.

Formula:

X' = (X - Xmin) / (Xmax - Xmin)

Where:

- X = Original value
- Xmin = Minimum value
- Xmax = Maximum value
- X' = Normalized value

In this practical, StandardScaler is used for scaling the selected numerical attributes.

## 4. ALGORITHM

1. Start.
2. Import Pandas, LabelEncoder and StandardScaler libraries.
3. Load the Titanic dataset using the read_csv() function.
4. Display the original dataset using head().
5. Check missing values using isnull().sum().
6. Replace missing Age and Fare values using their mean.
7. Replace missing Embarked values using its mode.
8. Remove unnecessary columns using the drop() function.
9. Apply Label Encoding to the Sex column.
10. Apply One-Hot Encoding to the Embarked column.
11. Select the numerical attributes Pclass, Age, SibSp, Parch and Fare.
12. Apply StandardScaler to the selected numerical attributes.
13. Display the dataset after encoding and scaling.
14. Check missing values again.
15. Separate input features X and target variable y.
16. Save the processed dataset using to_csv().
17. Stop.

## 5. DETAILED EXPLANATION OF EACH LINE OF CODE

### 1. Importing Libraries

`import pandas as pd`

This statement imports the Pandas library and gives it the short name pd. Pandas is used for reading, modifying and analysing datasets.

`from sklearn.preprocessing import LabelEncoder, StandardScaler`

This imports LabelEncoder and StandardScaler from Scikit-learn's preprocessing module.

- LabelEncoder converts categorical values into numerical labels.
- StandardScaler standardizes numerical attributes.

### 2. Loading the Dataset

`df = pd.read_csv(r"C:\Users\Tanuja\Downloads\titanic.csv")`

- read_csv() reads the CSV file.
- The given path specifies the location of the Titanic dataset.
- df stores the dataset in a Pandas DataFrame.
- The letter r represents a raw string and helps handle Windows file paths.

### 3. Displaying the Original Dataset

`print("Original Dataset:")`

Displays the heading Original Dataset.

`print(df.head())`

The head() function displays the first five rows of the dataset. It helps us understand the dataset structure and column names.

### 4. Checking Missing Values

`print(df.isnull().sum())`

- isnull() identifies missing values in every cell.
- It returns True for missing values and False for available values.
- sum() counts the missing values column-wise.

This step helps identify which columns require missing value treatment.

### 5. Handling Missing Numerical Values

`df["Age"] = df["Age"].fillna(df["Age"].mean())`

- df["Age"] selects the Age column.
- mean() calculates the average Age.
- fillna() replaces missing Age values with the calculated mean.
- The updated values are stored back in the Age column.

`df["Fare"] = df["Fare"].fillna(df["Fare"].mean())`

This statement performs the same operation on the Fare column by replacing missing values with the average Fare.

**Why is mean used?**

Mean is a simple method for replacing missing numerical values. However, it is sensitive to outliers. Median may be a better option when the data contains extreme values.

### 6. Handling Missing Categorical Values

`df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])`

- df["Embarked"] selects the Embarked column.
- mode() finds the most frequently occurring category.
- [0] selects the first mode.
- fillna() replaces missing values with that category.

Mode is suitable for categorical data because categories cannot be averaged.

### 7. Removing Unnecessary Columns

`df.drop(["PassengerId", "Name", "Ticket", "Cabin"], axis=1, inplace=True)`

This statement removes four columns from the dataset.

- PassengerId: Used only as a passenger identifier.
- Name: Contains raw text information.
- Ticket: Contains ticket identification information.
- Cabin: Contains a large proportion of missing values in the standard Titanic training dataset.

**Parameters:**

- drop(): Removes rows or columns.
- axis=1: Specifies column-wise deletion.
- inplace=True: Modifies the original DataFrame directly.

These columns are removed in this basic preprocessing implementation because they are not used as features here. Some of them could contain useful information after additional feature extraction, but that is outside the scope of this practical.

### 8. Applying Label Encoding

`le = LabelEncoder()`

Creates an object of the LabelEncoder class.

`df["Sex"] = le.fit_transform(df["Sex"])`

- fit() identifies the available categories.
- transform() converts categories into numerical labels.
- fit_transform() performs both operations together.

For the standard Titanic dataset, female is encoded as 0 and male as 1 because LabelEncoder sorts the category labels.

### 9. Applying One-Hot Encoding

`df = pd.get_dummies(df, columns=["Embarked"], dtype=int)`

- get_dummies() performs One-Hot Encoding.
- It creates separate binary columns for each Embarked category.
- dtype=int ensures that the generated columns contain integer values.

The Embarked column is replaced by columns such as Embarked_C, Embarked_Q and Embarked_S.

### 10. Selecting Numerical Attributes

`cols = ["Pclass", "Age", "SibSp", "Parch", "Fare"]`

This statement creates a list containing the numerical attributes selected for feature scaling.

- Pclass: Passenger ticket class.
- Age: Passenger age.
- SibSp: Number of siblings and spouses aboard.
- Parch: Number of parents and children aboard.
- Fare: Passenger ticket fare.

### 11. Applying Feature Scaling

`scaler = StandardScaler()`

Creates an object of the StandardScaler class.

`df[cols] = scaler.fit_transform(df[cols])`

This statement performs two operations:

- fit(): Calculates the mean and standard deviation of each selected feature.
- transform(): Standardizes the values using the calculated statistics.

Formula:

Z = (X - μ) / σ

The resulting features have approximately zero mean and unit standard deviation.

The Survived column is not included in scaling because it is the target variable and its original class labels must be preserved.

### 12. Checking Missing Values Again

`print(df.isnull().sum())`

Checks whether any missing values remain after preprocessing.

This helps verify whether the missing value treatment was successful.

### 13. Separating Input and Output Variables

`X = df.drop("Survived", axis=1)`

Stores all the input features in X after removing the target column.

`y = df["Survived"]`

Stores the Survived column in y.

In this dataset:

- 0 represents a passenger who did not survive.
- 1 represents a passenger who survived.

### 14. Saving the Cleaned Dataset

`df.to_csv(r"C:\Users\Tanuja\Downloads\cleaned_titanic_dataset.csv", index=False)`

- to_csv() saves the processed dataset as a CSV file.
- The specified path determines where the file is stored.
- index=False prevents an additional DataFrame index column from being saved.

## 6. IMPORTANT FUNCTIONS USED

| Function | Purpose |
|---|---|
| read_csv() | Loads a CSV dataset |
| head() | Displays first five rows |
| info() | Displays dataset information |
| isnull() | Identifies missing values |
| sum() | Counts missing values |
| mean() | Calculates arithmetic average |
| mode() | Finds most frequent value |
| fillna() | Replaces missing values |
| drop() | Removes unnecessary columns |
| fit_transform() | Learns and applies a transformation |
| get_dummies() | Performs One-Hot Encoding |
| StandardScaler() | Standardizes numerical features |
| to_csv() | Saves the processed dataset |
| shape | Returns the number of rows and columns |

## 7. EXPECTED OUTPUT

For the standard Titanic training dataset:

- Original dataset: 891 rows and 12 columns.
- After removing four columns: 891 rows and 8 columns.
- After One-Hot Encoding: 891 rows and 10 columns.
- Final processed dataset: 891 rows and 10 columns.
- Input feature matrix X: 891 rows and 9 columns.
- Target variable y: 891 values.

The final dataset should have no missing values if it contains the standard columns and values assumed by this implementation.

## 8. ORAL EXAMINATION – VIVA QUESTIONS AND ANSWERS

### Basic Conceptual Questions

**Q1. What is Feature Engineering?**

Feature Engineering is the process of selecting, creating and transforming data features to make them suitable for machine learning algorithms.

**Q2. Why is Feature Engineering required?**

It prepares raw data by handling missing values, converting categorical data and scaling numerical attributes.

**Q3. What is a feature?**

A feature is an individual input attribute or variable used by a machine learning model for prediction.

**Q4. What is a dataset?**

A dataset is a collection of related data arranged in rows and columns.

**Q5. What is a DataFrame?**

A DataFrame is a two-dimensional tabular data structure provided by Pandas.

### Questions on Missing Values

**Q6. What are missing values?**

Missing values are data entries that are not available or recorded in a dataset. They are commonly represented by NaN.

**Q7. What is data imputation?**

Data imputation is the process of replacing missing values with suitable estimated values.

**Q8. What are the different methods of handling missing values?**

Mean imputation, median imputation, mode imputation, row deletion and column deletion.

**Q9. Why is mean used for numerical values?**

Mean replaces missing numerical values with the average of the available observations.

**Q10. Why is mode used for categorical values?**

Mode identifies the most frequently occurring category and can be used to replace missing categorical values.

### Questions on Encoding

**Q11. What is categorical data?**

Categorical data represents labels or groups, such as gender, city or department.

**Q12. What is Label Encoding?**

Label Encoding converts categorical values into numerical labels by assigning a unique integer to each category.

**Q13. What is One-Hot Encoding?**

One-Hot Encoding converts categories into separate binary columns containing 0 and 1.

**Q14. What is the difference between Label Encoding and One-Hot Encoding?**

Label Encoding assigns numerical labels, while One-Hot Encoding creates separate binary columns for categories.

**Q15. Why is One-Hot Encoding used for Embarked?**

Because Embarked contains nominal categories without any natural order. One-Hot Encoding prevents an artificial ordering among these categories.

### Questions on Feature Scaling

**Q16. What is Feature Scaling?**

Feature Scaling is the process of transforming numerical attributes into a comparable scale.

**Q17. What is StandardScaler?**

StandardScaler is a Scikit-learn technique that transforms numerical data to approximately zero mean and unit standard deviation.

**Q18. What is the formula of StandardScaler?**

Z = (X - μ) / σ

**Q19. What is normalization?**

Normalization commonly transforms numerical values into a fixed range, such as 0 to 1.

**Q20. Does StandardScaler remove outliers?**

No. It changes the scale of the numerical values but does not remove outliers.

## 9. DIFFICULT EXTERNAL EXAMINER QUESTIONS AND ANSWERS

**Q21. Why did you select the Titanic dataset for this practical?**

The Titanic dataset contains numerical attributes, categorical attributes and missing values. Therefore, all three required feature engineering techniques can be demonstrated on a single dataset.

**Q22. Why is median sometimes preferred over mean?**

Median is less affected by extreme values or outliers. When numerical data is highly skewed, median can be a more representative imputation method.

**Q23. Why should categorical data not be replaced using mean?**

Categorical values represent labels rather than numerical measurements. Therefore, calculating their arithmetic mean is not meaningful.

**Q24. What is the difference between nominal and ordinal data?**

Nominal data contains categories without a natural order, such as city or gender. Ordinal data contains categories with a meaningful order, such as low, medium and high.

**Q25. Does StandardScaler make the dataset normally distributed?**

No. It standardizes the mean and standard deviation but does not guarantee a normal distribution.

**Q26. What is Data Leakage?**

Data Leakage occurs when information from the testing dataset enters the model training process, resulting in misleading evaluation results.

**Q27. How can Data Leakage be prevented during feature scaling?**

First divide the dataset into training and testing sets. Fit the scaler only on training data and use the same fitted scaler to transform testing data.

**Q28. What is the difference between fit(), transform() and fit_transform()?**

- fit(): Learns parameters such as mean and standard deviation.
- transform(): Applies the learned transformation.
- fit_transform(): Performs both operations together.

**Q29. Why is the Survived column not scaled?**

Survived is the target variable containing class labels 0 and 1. Scaling it would unnecessarily change the original class labels.

**Q30. Why is the Cabin column removed?**

The Cabin column contains a large proportion of missing values in the standard Titanic training dataset. It is removed in this basic implementation to avoid unreliable imputation.

**Q31. Which machine learning algorithms benefit from feature scaling?**

KNN, SVM and Logistic Regression generally benefit from feature scaling because they are sensitive to feature magnitudes or distances.

**Q32. Do Decision Trees require feature scaling?**

Generally, no. Decision Trees use feature-based splitting thresholds and are usually unaffected by different numerical scales.

**Q33. What happens if missing values are not handled?**

Some preprocessing and machine learning algorithms may fail or produce unreliable results when missing values are present.

**Q34. What is the difference between Feature Selection and Feature Extraction?**

Feature Selection selects important existing features from a dataset, whereas Feature Extraction transforms existing data into new features.

**Q35. Can we apply StandardScaler to categorical variables directly?**

No. Categorical values must first be converted into a suitable numerical representation before numerical scaling can be applied.

## 10. PRACTICAL EXAMINATION – CODING QUESTIONS

**Q1. How do you load a CSV file?**

`df = pd.read_csv("file.csv")`

**Q2. How do you display the first five rows?**

`df.head()`

**Q3. How do you check missing values?**

`df.isnull().sum()`

**Q4. How do you calculate the mean of Age?**

`df["Age"].mean()`

**Q5. How do you calculate the median of Age?**

`df["Age"].median()`

**Q6. How do you calculate the mode of Embarked?**

`df["Embarked"].mode()[0]`

**Q7. How do you replace missing values?**

`df["Age"] = df["Age"].fillna(df["Age"].mean())`

**Q8. How do you remove a column?**

`df.drop("Name", axis=1)`

**Q9. How do you apply Label Encoding?**

`le.fit_transform(df["Sex"])`

**Q10. How do you apply One-Hot Encoding?**

`pd.get_dummies(df, columns=["Embarked"])`

**Q11. How do you perform Feature Scaling?**

`scaler.fit_transform(df[cols])`

**Q12. How do you check the dimensions of a dataset?**

`df.shape`

**Q13. How do you save the cleaned dataset?**

`df.to_csv("cleaned_dataset.csv", index=False)`

## 11. DATASET-RELATED QUESTIONS

**Q1. What is the Titanic dataset?**

It is a passenger dataset containing information about passengers aboard the Titanic ship and their survival status.

**Q2. What is the main purpose of the Titanic dataset?**

It is commonly used to analyse passenger information and predict whether passengers survived or not.

**Q3. What does the Survived column represent?**

It represents the survival status of passengers. 0 means did not survive and 1 means survived.

**Q4. What does Pclass represent?**

Pclass represents the passenger's ticket class: 1st, 2nd or 3rd class.

**Q5. What does SibSp represent?**

SibSp represents the number of siblings and spouses travelling with a passenger aboard the Titanic.

**Q6. What does Parch represent?**

Parch represents the number of parents and children travelling with a passenger aboard the Titanic.

**Q7. What does Embarked represent?**

Embarked represents the port from which a passenger boarded the ship.

- C = Cherbourg
- Q = Queenstown
- S = Southampton

**Q8. What does Fare represent?**

Fare represents the ticket fare paid by a passenger.

**Q9. Why is PassengerId removed?**

It is primarily an identification number and is not used as an input feature in this basic practical.

**Q10. Why are Age and Fare considered numerical attributes?**

Because they contain measurable numerical values.

**Q11. Why are Sex and Embarked considered categorical attributes?**

Because they represent categories rather than continuous numerical measurements.

**Q12. What is the target variable in the Titanic dataset?**

Survived is the target variable.

## 12. IMPORTANT FORMULAS

**Mean:**

Mean = Sum of all values / Number of values

**Standardization:**

Z = (X - μ) / σ

**Normalization:**

X' = (X - Xmin) / (Xmax - Xmin)

**Remember:**

- Mean: Average value.
- Median: Middle value.
- Mode: Most frequent value.
- Label Encoding: Converts categories into numerical labels.
- One-Hot Encoding: Creates binary columns.
- StandardScaler: Standardizes numerical attributes.
- Data Leakage: Unwanted use of test information during training.
```
