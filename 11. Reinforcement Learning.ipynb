````python
import random

random.seed(42)

q_table = [[0, 0] for i in range(5)]

learning_rate = 0.1
gamma = 0.9
epsilon = 1.0
episodes = 1000

for episode in range(episodes):
    state = 0

    for step in range(10):

        if random.random() < epsilon:
            action = random.randint(0, 1)
        else:
            action = q_table[state].index(max(q_table[state]))

        if action == 0:
            next_state = max(0, state - 1)
        else:
            next_state = min(4, state + 1)

        if next_state == 4:
            reward = 10
            done = True
        else:
            reward = -1
            done = False

        if done:
            q_table[state][action] += learning_rate * (
                reward - q_table[state][action]
            )
        else:
            q_table[state][action] += learning_rate * (
                reward + gamma * max(q_table[next_state])
                - q_table[state][action]
            )

        state = next_state

        if done:
            break

    epsilon = max(0.01, epsilon * 0.995)

print("Q-Table:")
for row in q_table:
    print([round(value, 2) for value in row])

print("\nOptimal Policy:")

for state in range(4):
    action = q_table[state].index(max(q_table[state]))

    if action == 0:
        print("State", state, ": Left")
    else:
        print("State", state, ": Right")

print("\nTesting Learned Policy:")

state = 0
path = [state]

for step in range(10):
    action = q_table[state].index(max(q_table[state]))

    if action == 0:
        state = max(0, state - 1)
    else:
        state = min(4, state + 1)

    path.append(state)

    if state == 4:
        break

print("Path:", path)

if state == 4:
    print("Goal Reached Successfully!")
else:
    print("Goal Not Reached!")
````

````python
%pip install numpy matplotlib
# PRACTICAL 11: REINFORCEMENT LEARNING

## 1. Aim

To evaluate the performance of Reinforcement Learning using the Q-Learning algorithm for a simple decision-making problem, focusing on exploration–exploitation strategies and learning the optimal policy.

## 2. Environment Used

**Environment:** Simple 5-State Decision-Making Environment.

The environment contains five states numbered from 0 to 4.

| State | Description |
|---|---|
| 0 | Starting position |
| 1 | Position 2 |
| 2 | Position 3 |
| 3 | Position 4 |
| 4 | Goal state |

**Available actions:**

- Left (0): Move one position to the left.
- Right (1): Move one position to the right.

**Reward system:**

| Condition | Reward |
|---|---:|
| Reaching the goal state (4) | +10 |
| Every other move | -1 |

The agent starts from state 0 and learns how to reach state 4 by selecting suitable actions.

## 3. Theory

### What is Reinforcement Learning?

Reinforcement Learning (RL) is a type of Machine Learning in which an agent learns how to make decisions by interacting with an environment.

The agent performs actions and receives rewards or penalties based on its actions.

The main objective is to learn a strategy that maximizes the total reward over time.

**Example:** A robot learns to reach a destination by trying different movements. It receives a reward for reaching the destination and a penalty for taking unnecessary steps.

### Important Elements of Reinforcement Learning

**1. Agent**

The agent is the learner or decision-maker that interacts with the environment.

In this practical, the agent learns how to move from state 0 to state 4.

**2. Environment**

The environment is the system in which the agent performs actions.

Here, the environment consists of five states.

**3. State**

A state represents the current position or situation of the agent.

In this practical, states are numbered from 0 to 4.

**4. Action**

An action is a decision taken by the agent.

Here, the agent can choose Left or Right.

**5. Reward**

A reward is feedback received by the agent after performing an action.

A positive reward encourages desirable actions, while a negative reward discourages undesirable actions.

**6. Policy**

A policy is a strategy that tells the agent which action to select in a particular state.

The objective of Q-Learning is to learn an effective policy.

### What is Q-Learning?

Q-Learning is a model-free, value-based Reinforcement Learning algorithm.

It learns the quality of different actions in different states without requiring prior knowledge of the environment's transition model.

It stores the learned values in a Q-table.

The Q-table contains the expected future rewards for taking a particular action from a particular state and following the learned policy afterward.

The agent selects actions based on these Q-values and gradually improves its decisions through repeated interactions.

### What is a Q-Table?

A Q-table is a table that stores Q-values for every state-action pair.

In this practical:

- Number of states = 5.
- Number of actions = 2.
- Q-table size = 5 × 2.

Initially, all Q-values are set to zero.

As the agent interacts with the environment, the Q-values are updated based on the rewards received.

### Q-Learning Formula

The Q-Learning update equation is:

Q(s,a) = Q(s,a) + α [r + γ max Q(s',a') − Q(s,a)]

Where:

- Q(s,a) = Current Q-value.
- α (Alpha) = Learning rate.
- r = Immediate reward.
- γ (Gamma) = Discount factor.
- s = Current state.
- a = Current action.
- s' = Next state.
- a' = Possible action in the next state.
- max Q(s',a') = Maximum estimated future reward from the next state.

For a terminal state, the future reward term is not included because the episode ends.

### Important Parameters in Q-Learning

**1. Learning Rate (Alpha)**

The learning rate controls how much new information affects the existing Q-value.

In this practical:

α = 0.1

A smaller value means slower learning, whereas a larger value gives more importance to new information.

**2. Discount Factor (Gamma)**

The discount factor determines the importance of future rewards compared with immediate rewards.

In this practical:

γ = 0.9

A value close to 1 gives more importance to future rewards.

**3. Epsilon (Exploration Rate)**

Epsilon controls the probability of selecting a random action.

Initially, epsilon is set to 1.0, meaning the agent explores randomly.

As training progresses, epsilon decreases, allowing the agent to use its learned knowledge more frequently.

## 4. Exploration and Exploitation

Exploration and exploitation are two important strategies in Reinforcement Learning.

### 1. Exploration

Exploration means trying different actions to discover which actions provide better rewards.

For example, the agent randomly selects Left or Right to understand the environment.

Exploration helps the agent discover new possibilities.

### 2. Exploitation

Exploitation means selecting the action with the highest known Q-value to obtain the maximum expected reward.

For example, if moving Right has a higher Q-value than moving Left, the agent selects Right.

Exploitation uses the knowledge already learned by the agent.

### Epsilon-Greedy Strategy

The epsilon-greedy strategy balances exploration and exploitation.

- With probability epsilon, the agent selects a random action.
- With probability 1 − epsilon, the agent selects the action with the highest Q-value.

Initially, the agent explores more. As epsilon decreases, it gradually exploits the learned knowledge.

In this practical, epsilon starts at 1.0 and decreases by multiplying it by 0.995 after each episode, subject to a minimum value of 0.01.

### What is an Episode?

An episode is one complete attempt by the agent to interact with the environment.

In this practical, every episode starts from state 0 and continues until the agent reaches the goal or completes the maximum number of steps.

A total of 1000 training episodes are performed.

### What is an Optimal Policy?

An optimal policy is a strategy that selects the best action in each state to maximize the expected total reward.

In this practical, the learned policy should guide the agent from state 0 towards state 4.

The best action in each state is selected using the highest Q-value.

## 5. Algorithm

1. Import the Random module.
2. Set a fixed random seed.
3. Initialize the Q-table with zeros for five states and two actions.
4. Set the learning rate, discount factor, exploration rate and number of episodes.
5. Start the training process for 1000 episodes.
6. Initialize the agent at state 0 at the beginning of each episode.
7. Select an action using the epsilon-greedy strategy.
8. Calculate the next state based on the selected action.
9. Assign a reward of +10 when the goal is reached; otherwise, assign -1.
10. Update the Q-table using the Q-Learning formula.
11. Continue until the goal is reached or the maximum number of steps is completed.
12. Reduce epsilon after every episode.
13. Display the learned Q-table.
14. Generate the optimal policy using the highest Q-value for each state.
15. Test the learned policy and display the path followed by the agent.

## 6. Python Program

### Installation

No external installation is required because the program uses Python's built-in Random module.

### Main Program

```python
import random

random.seed(42)

q_table = [[0, 0] for i in range(5)]

learning_rate = 0.1
gamma = 0.9
epsilon = 1.0
episodes = 1000

for episode in range(episodes):
    state = 0

    for step in range(10):

        if random.random() < epsilon:
            action = random.randint(0, 1)
        else:
            action = q_table[state].index(max(q_table[state]))

        if action == 0:
            next_state = max(0, state - 1)
        else:
            next_state = min(4, state + 1)

        if next_state == 4:
            reward = 10
            done = True
        else:
            reward = -1
            done = False

        if done:
            q_table[state][action] += learning_rate * (
                reward - q_table[state][action]
            )
        else:
            q_table[state][action] += learning_rate * (
                reward + gamma * max(q_table[next_state])
                - q_table[state][action]
            )

        state = next_state

        if done:
            break

    epsilon = max(0.01, epsilon * 0.995)

print("Q-Table:")
for row in q_table:
    print([round(value, 2) for value in row])

print("\nOptimal Policy:")

for state in range(4):
    action = q_table[state].index(max(q_table[state]))

    if action == 0:
        print("State", state, ": Left")
    else:
        print("State", state, ": Right")

print("\nTesting Learned Policy:")

state = 0
path = [state]

for step in range(10):
    action = q_table[state].index(max(q_table[state]))

    if action == 0:
        state = max(0, state - 1)
    else:
        state = min(4, state + 1)

    path.append(state)

    if state == 4:
        break

print("Path:", path)

if state == 4:
    print("Goal Reached Successfully!")
else:
    print("Goal Not Reached!")
```

## 7. Line-by-Line Explanation of the Program

### A. Importing and Initializing

**Line 1:**
```python
import random
```
Imports Python's built-in Random module. It is used to select random actions during exploration.

**Line 2:**
```python
random.seed(42)
```
Sets a fixed random seed so that the random sequence can be reproduced.

**Line 3:**
```python
q_table = [[0, 0] for i in range(5)]
```
Creates a Q-table containing five rows and two columns.

- Five rows represent five states.
- Two columns represent Left and Right actions.
- Initially, all Q-values are zero.

**Line 4:**
```python
learning_rate = 0.1
```
Sets the learning rate to 0.1. It controls how much new information changes the existing Q-value.

**Line 5:**
```python
gamma = 0.9
```
Sets the discount factor to 0.9. It determines the importance of future rewards.

**Line 6:**
```python
epsilon = 1.0
```
Initializes the exploration rate to 1.0. This means that the agent initially selects random actions.

**Line 7:**
```python
episodes = 1000
```
Specifies that the agent will undergo 1000 training episodes.

### B. Training the Agent

**Line 8:**
```python
for episode in range(episodes):
```
Starts the training loop and repeats the learning process 1000 times.

**Line 9:**
```python
state = 0
```
Sets the initial state to 0 at the beginning of every episode.

**Line 10:**
```python
for step in range(10):
```
Allows the agent to take a maximum of 10 steps in one episode.

### C. Exploration and Exploitation

**Lines 11–12:**
```python
if random.random() < epsilon:
    action = random.randint(0, 1)
```
Checks whether the agent should explore.

- `random.random()` generates a random number between 0 and 1.
- If the number is smaller than epsilon, the agent chooses a random action.
- `random.randint(0, 1)` selects either Left (0) or Right (1).

**Lines 13–14:**
```python
else:
    action = q_table[state].index(max(q_table[state]))
```
Executes exploitation when the random number is not smaller than epsilon.

- `q_table[state]` accesses the Q-values of the current state.
- `max()` finds the highest Q-value.
- `index()` returns the action corresponding to that highest value.

Thus, the agent selects the action it currently considers best.

### D. Calculating the Next State

**Lines 15–17:**
```python
if action == 0:
    next_state = max(0, state - 1)
else:
    next_state = min(4, state + 1)
```

Determines the next state based on the selected action.

- If action is 0, the agent moves Left.
- If action is 1, the agent moves Right.
- `max(0, state - 1)` prevents the agent from moving below state 0.
- `min(4, state + 1)` prevents the agent from moving beyond state 4.

### E. Assigning Rewards

**Lines 18–20:**
```python
if next_state == 4:
    reward = 10
    done = True
```
Checks whether the agent has reached the goal state.

If the next state is 4, the agent receives a reward of +10 and the episode is marked as completed.

**Lines 21–23:**
```python
else:
    reward = -1
    done = False
```
If the goal is not reached, the agent receives a penalty of -1 and continues the episode.

### F. Updating the Q-Table

**Lines 24–27:**
```python
if done:
    q_table[state][action] += learning_rate * (
        reward - q_table[state][action]
    )
```
Updates the Q-value when the agent reaches the terminal state.

Since the episode has ended, no future reward is considered.

**Lines 28–31:**
```python
else:
    q_table[state][action] += learning_rate * (
        reward + gamma * max(q_table[next_state])
        - q_table[state][action]
    )
```
Updates the Q-value when the agent has not reached the goal.

It uses the Q-Learning formula:

Q(s,a) = Q(s,a) + α [r + γ max Q(s',a') − Q(s,a)]

The update considers the immediate reward and the maximum estimated future reward.

### G. Moving to the Next State

**Line 32:**
```python
state = next_state
```
Updates the current state with the newly calculated next state.

**Lines 33–34:**
```python
if done:
    break
```
Stops the current episode when the goal is reached.

### H. Reducing Exploration

**Line 35:**
```python
epsilon = max(0.01, epsilon * 0.995)
```
Reduces the exploration rate after each episode.

- Multiplying epsilon by 0.995 gradually decreases exploration.
- `max(0.01, ...)` ensures that epsilon never becomes smaller than 0.01.

This allows the agent to gradually shift from exploration to exploitation.

### I. Displaying the Q-Table

**Line 36:**
```python
print("Q-Table:")
```
Displays a heading for the learned Q-table.

**Lines 37–38:**
```python
for row in q_table:
    print([round(value, 2) for value in row])
```
Displays each row of the Q-table.

- The `for` loop accesses each state.
- `round(value, 2)` displays each Q-value up to two decimal places.

### J. Displaying the Optimal Policy

**Line 39:**
```python
print("\nOptimal Policy:")
```
Displays the optimal policy heading.

**Line 40:**
```python
for state in range(4):
```
Checks the learned policy for states 0, 1, 2 and 3.

The goal state is excluded because no further action is needed after reaching it.

**Line 41:**
```python
action = q_table[state].index(max(q_table[state]))
```
Selects the action with the highest Q-value for the current state.

**Lines 42–45:**
```python
if action == 0:
    print("State", state, ": Left")
else:
    print("State", state, ": Right")
```
Displays the best learned action for the current state.

### K. Testing the Learned Policy

**Line 46:**
```python
print("\nTesting Learned Policy:")
```
Displays the testing heading.

**Line 47:**
```python
state = 0
```
Starts the testing process from state 0.

**Line 48:**
```python
path = [state]
```
Creates a list to store the path followed by the agent.

**Line 49:**
```python
for step in range(10):
```
Allows the agent to take a maximum of 10 steps during testing.

**Line 50:**
```python
action = q_table[state].index(max(q_table[state]))
```
Selects the best learned action using the highest Q-value. Unlike the training phase, this testing phase does not use random exploration.

**Lines 51–54:**
```python
if action == 0:
    state = max(0, state - 1)
else:
    state = min(4, state + 1)
```
Updates the state according to the selected action while keeping the agent inside the environment boundaries.

**Line 55:**
```python
path.append(state)
```
Adds the new state to the path list.

**Lines 56–57:**
```python
if state == 4:
    break
```
Stops testing when the agent reaches the goal state.

**Line 58:**
```python
print("Path:", path)
```
Displays the path followed by the agent.

**Lines 59–62:**
```python
if state == 4:
    print("Goal Reached Successfully!")
else:
    print("Goal Not Reached!")
```
Checks whether the agent successfully reached state 4 and displays the corresponding message.

## 8. Expected Output

The program displays the following:

1. Learned Q-table containing Q-values for each state-action pair.
2. Optimal policy showing Left or Right for each state.
3. Path followed by the agent during testing.
4. Message indicating whether the goal was successfully reached.

**Expected learned path:**

`[0, 1, 2, 3, 4]`

The agent learns to move Right from the starting position until it reaches the goal.

The exact Q-values can vary depending on the training process, but with the fixed random seed and the given environment, the results are reproducible under the same Python execution conditions.

## 9. Important Viva Questions and Answers

**Q1. What is Reinforcement Learning?**

Reinforcement Learning is a type of Machine Learning in which an agent learns by interacting with an environment and receiving rewards or penalties.

**Q2. What is Q-Learning?**

Q-Learning is a model-free, value-based Reinforcement Learning algorithm that learns the best action for each state using a Q-table.

**Q3. What is an agent?**

An agent is the learner or decision-maker that interacts with the environment.

**Q4. What is an environment?**

An environment is the system in which the agent performs actions and receives rewards.

**Q5. What is a state?**

A state represents the current position or condition of the agent.

**Q6. What is an action?**

An action is a decision taken by the agent to change its state.

**Q7. What is a reward?**

A reward is feedback received by the agent after performing an action.

**Q8. What is a Q-table?**

A Q-table stores the estimated value of every state-action pair.

**Q9. What is exploration?**

Exploration means trying different actions to discover which actions provide better rewards.

**Q10. What is exploitation?**

Exploitation means selecting the action with the highest known Q-value.

**Q11. What is the epsilon-greedy strategy?**

It is a strategy in which the agent selects random actions with probability epsilon and the best-known action with probability 1 − epsilon.

**Q12. What is the learning rate?**

The learning rate determines how much new information affects the existing Q-value.

**Q13. What is the discount factor?**

The discount factor determines the importance of future rewards.

**Q14. What is an episode?**

An episode is one complete attempt by the agent to interact with the environment, beginning from the starting state and ending at a terminal condition or maximum step limit.

**Q15. What is an optimal policy?**

An optimal policy is a strategy that selects the best action in each state to maximize expected total rewards.

## 10. Advanced External Viva Questions

**Q1. Why is Q-Learning called model-free?**

It is called model-free because it does not require prior knowledge of the environment's transition probabilities or reward model. It learns from interaction.

**Q2. What happens if epsilon is always 1?**

The agent always selects random actions and does not use its learned Q-values for exploitation.

**Q3. What happens if epsilon is always 0?**

The agent always selects the currently best-known action and never explores new actions. This may prevent it from discovering better actions.

**Q4. Why is epsilon gradually reduced?**

Initially, exploration helps the agent discover the environment. Later, reducing epsilon encourages the agent to exploit the knowledge it has learned.

**Q5. What happens if the learning rate is too high?**

The agent may give too much importance to recent information, which can make learning unstable.

**Q6. What happens if the discount factor is 0?**

The agent considers only immediate rewards and ignores future rewards.

**Q7. What happens if the discount factor is close to 1?**

The agent gives greater importance to future rewards.

**Q8. What is the difference between Q-Learning and supervised learning?**

| Q-Learning | Supervised Learning |
|---|---|
| Learns through interaction and rewards | Learns using labelled training data |
| Does not require predefined correct action labels | Uses known target labels |
| Learns action values and a policy | Learns a mapping from inputs to outputs |

**Q9. Why is a negative reward assigned for every non-goal move?**

It encourages the agent to reach the goal using fewer unnecessary steps.

**Q10. Why is the Q-table initially filled with zeros?**

Zero initialization provides a simple starting value for learning. The Q-values are updated as the agent interacts with the environment.

**Q11. What is a terminal state?**

A terminal state is a state in which the current episode ends. In this practical, state 4 is the terminal goal state.

**Q12. Is Q-Learning suitable for large state spaces?**

A traditional Q-table can become very large for high-dimensional or continuous state spaces. Methods such as Deep Q-Networks can be used in such cases.

## 11. Practical-Specific Viva Questions

**Q1. How many states are used in this practical?**

Five states, numbered from 0 to 4.

**Q2. What is the starting state?**

State 0.

**Q3. What is the goal state?**

State 4.

**Q4. How many actions are available?**

Two actions: Left and Right.

**Q5. What reward is given for reaching the goal?**

A positive reward of +10.

**Q6. What penalty is given for other moves?**

A reward of -1.

**Q7. How many training episodes are used?**

1000 episodes.

**Q8. What is the initial exploration rate?**

1.0.

**Q9. What is the minimum exploration rate?**

0.01.

**Q10. What is the purpose of the testing phase?**

It checks whether the learned policy can guide the agent from state 0 to state 4.

## 12. Important Points to Remember for Viva

- Practical: Reinforcement Learning.
- Algorithm: Q-Learning.
- Learning type: Reinforcement Learning.
- Environment: Five-state decision-making problem.
- Starting state: 0.
- Goal state: 4.
- Available actions: Left and Right.
- Q-table size: 5 × 2.
- Learning rate: 0.1.
- Discount factor: 0.9.
- Initial epsilon: 1.0.
- Minimum epsilon: 0.01.
- Training episodes: 1000.
- Goal reward: +10.
- Other movement reward: -1.
- Exploration strategy: Epsilon-Greedy.
- Output: Learned Q-table, optimal policy and path to the goal.
````
